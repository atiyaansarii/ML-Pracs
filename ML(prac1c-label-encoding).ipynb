```python
import pandas as pd

from sklearn.preprocessing import LabelEncoder
from sklearn.preprocessing import StandardScaler
from sklearn.preprocessing import Binarizer

data = {
    'Customer': ['C101', 'C102', 'C103', 'C104', 'C105', 'C106', 'C107', 'C108'],
    'Gender': ['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Age': [25, 32, 28, 45, 35, 29, 40, 23],
    'Salary': [30000, 55000, 40000, 80000, 60000, 45000, 70000, 25000],
    'PurchaseAmopunt': [2000, 5000, 3000, 9000, 7000, 4000, 8500, 1500]
}

df = pd.DataFrame(data)
print("\nOriginal DataFrame:")
print(df)

encoder = LabelEncoder()
df['Gender_Encoded'] = encoder.fit_transform(df['Gender'])
print("\n after Label Encoding:")
print(df[['Customer', 'Gender', 'Gender_Encoded']])

scaler = StandardScaler()
df['Salary_Scaled'] = scaler.fit_transform(df[['Salary']])
print("\n after Standard Scaling:")
print(df[['Customer', 'Gender', 'Gender_Encoded']])

scaler = StandardScaler()
df[['Age_Scaled', 'Salary_Scaled']] = scaler.fit_transform(df[['Age', 'Salary']])
print("\n after Standard Scaling:")
print(df[['Customer', 'Age', 'Age_Scaled', 'Salary', 'Salary_Scaled']])

binarizer = Binarizer(threshold=5000)
df['Purchase_Binary'] = binarizer.fit_transform(df[['PurchaseAmopunt']])
print("\n after Binarizer Scaling:")
print(df[['Customer', 'PurchaseAmopunt', 'Purchase_Binary']])

print("\n Final Dataset")
print(df)

print("\n Practical Compeleted Successfully")
```
